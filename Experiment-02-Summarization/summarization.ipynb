```python
article = """
Deep learning has significantly advanced the field of natural language processing.
Sequence to sequence models with attention mechanisms are widely used for tasks such as
machine translation, text summarization, and question answering. These models consist of
an encoder that processes the input sequence and a decoder that generates the output sequence.
Recently, transformer-based architectures have further improved performance across many NLP tasks.
"""
```

```python
import nltk
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(article)

extractive_summary = " ".join(sentences[:2])

print("EXTRACTIVE SUMMARY:\n")
print(extractive_summary)
```

```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)
```

```python
abstractive_summary = summarizer(
    article,
    max_length=60,
    min_length=25,
    do_sample=False
)

print("ABSTRACTIVE SUMMARY:\n")
print(abstractive_summary[0]['summary_text'])
```

```python
import evaluate

rouge = evaluate.load("rouge")

results = rouge.compute(
    predictions=[abstractive_summary[0]['summary_text']],
    references=[extractive_summary]
)

print(results)
```
