```python
from transformers import pipeline

tts = pipeline("text-to-speech", model="suno/bark-small")
```

```python
text = "Deep learning is transforming artificial intelligence and modern technology."
```

```python
speech = tts(text)
```

```python
import soundfile as sf

sf.write("output.wav", speech["audio"].flatten(), speech["sampling_rate"], subtype='FLOAT')
```
