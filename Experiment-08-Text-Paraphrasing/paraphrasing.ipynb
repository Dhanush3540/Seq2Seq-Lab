```python
from transformers import pipeline

paraphraser = pipeline(
    "text2text-generation",
    model="ramsrigouthamg/t5_paraphraser"
)
```

```python
sentences = [
    "Deep learning models are used in many NLP tasks.",
    "Sequence to sequence models generate output text.",
    "Artificial intelligence is transforming technology."
]
```

```python
for s in sentences:
    text = "paraphrase: " + s + " </s>"

    result = paraphraser(
        text,
        max_length=60,
        num_beams=5,
        num_return_sequences=1
    )

    print("Original  :", s)
    print("Paraphrase:", result[0]["generated_text"])
    print("-" * 50)
```
