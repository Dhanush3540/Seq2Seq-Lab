```python
from transformers import T5Tokenizer, T5ForConditionalGeneration

model_name = "t5-small"

tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)
```

```python
context = """
Sequence to sequence models are widely used in natural language processing.
They consist of an encoder and a decoder.
Applications include machine translation, summarization, and question answering.
"""

question = "What are sequence to sequence models used for?"
```

```python
input_text = f"question: {question}  context: {context}"

inputs = tokenizer(
    input_text,
    return_tensors="pt",
    truncation=True
)
```

```python
outputs = model.generate(
    inputs.input_ids,
    max_length=50
)

answer = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Question:", question)
print("Answer:", answer)
```
