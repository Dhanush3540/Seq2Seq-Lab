```python
from transformers import pipeline

grammar_corrector = pipeline(
    "text2text-generation",
    model="vennify/t5-base-grammar-correction"
)
```

```python
sentences = [
    "He go to school yesterday.",
    "She dont like playing games.",
    "I has completed the assignments.",
    "They is working on the project."
]
```

```python
for s in sentences:
    corrected = grammar_corrector(s, max_length=64)[0]["generated_text"]
    print("Original :", s)
    print("Corrected:", corrected)
    print("-" * 40)
```
